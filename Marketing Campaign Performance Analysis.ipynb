# Exploratory Data Analysis (EDA) & **Visualization** **bold text**

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

Scatter Plot (Acquisition_Cost vs ROI)

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Acquisition_Cost', y='ROI')
plt.title('Acquisition Cost vs ROI')
plt.show()
```

# **Bar Chart Average Conversion Rate by Channel & Campaign Type**

```python
plt.figure(figsize=(10, 6))
sns.barplot(data=df, x='Channel_Used', y='Conversion_Rate', hue='Campaign_Type', errorbar=None)
plt.title('Avg Conversion Rate by Channel & Campaign Type')
plt.show()
```

# **Box Plot Engagement Score across Campaign Type**

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='Campaign_Type', y='Engagement_Score', hue='Campaign_Type', palette='Set2', legend=False)
plt.title('Engagement Score Distribution by Campaign Type')
plt.show()
```

# **Bar Chart Average ROI by Company**

```python
plt.figure(figsize=(10, 5))
sns.barplot(data=df, x='Company', y='ROI', ci=None, palette='Set2')
plt.title('Average ROI by Company')
plt.show()
```

# **Heatmap Correlation: Engagement Score vs Conversion Rate**

```python
plt.figure(figsize=(6, 4))
sns.heatmap(df[['Engagement_Score', 'Conversion_Rate']].corr(), annot=True, cmap='coolwarm')
plt.title('Correlation: Engagement Score vs Conversion Rate')
plt.show()
```

# **Custome**r **Segmentation**

# Count Plot (Target_Audience Distribution)

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='Target_Audience',palette='Set2')
plt.title('Target Audience Distribution')
plt.show()
```

**Bar Chart Highest Conversion_Rate by Customer_Segment for each Language**

```python
plt.figure(figsize=(12, 6))
sns.barplot(data=df, x='Language', y='Conversion_Rate', hue='Customer_Segment', ci=None)
plt.title('Conversion Rate by Customer Segment per Language')
plt.show()
```

# Plot Acquisition Cost by Customer Segment & Channel

```python
plt.figure(figsize=(12, 6))
sns.boxplot(data=df, x='Customer_Segment', y='Acquisition_Cost', hue='Channel_Used')
plt.title('Acquisition Cost by Customer Segment and Channel')
plt.show()
```

# Bar Chart Average Conversion_Rate by Language

```python
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x='Language', y='Conversion_Rate',ci=None,palette='Set2',legend=False,errorbar=None)
plt.title('Average Conversion Rate by Language')
plt.show()
```

# Channel Effectiveness

Bar Chart (Engagement_Score by Channel segmented by Campaign_Type)

```python
plt.figure(figsize=(10, 6))
sns.barplot(data=df, x='Channel_Used', y='Engagement_Score', hue='Campaign_Type', errorbar=None)
plt.title('Engagement Score by Channel & Campaign Type')
plt.show()
```

# **Pie Chart (Distribution of Total ROI across Channels)**

```python
ROI_by_channel = df.groupby('Channel_Used')['ROI'].sum()
plt.figure(figsize=(7, 7))
plt.pie(ROI_by_channel, labels=ROI_by_channel.index, autopct='%1.1f%%')
plt.title('Total ROI Distribution across Channels')
plt.show()
```

# **Scatter Plot Clicks vs Impressions for each Channel **

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Impressions', y='Clicks', hue='Channel_Used')
plt.title('Clicks vs Impressions by Channel')
plt.show()
```

# **Time-Based Analysis**

**Histogram** (Distribution of Duration)

```python
plt.figure(figsize=(8, 5))
sns.histplot(df['Duration'], kde=True,color='purple')
plt.title('Distribution of Campaign Duration')
plt.show()
```

# **Line Chart Conversion Rate over Date for each Company**

```python
plt.figure(figsize=(12, 6))
sns.lineplot(data=df, x='Date', y='Conversion_Rate', hue='Company')
plt.title('Conversion Rate Trend Over Time by Company')
plt.show()
```

# **Line Chart Trend of Engagement Score over Date**

```python
plt.figure(figsize=(10, 5))
sns.lineplot(data=df, x='Date', y='Engagement_Score',color='green')
plt.title('Engagement Score Trend Over Time')
plt.show()
```

# **Geographic Analysis**

# Bar Chart (Highest Acquisition_Cost by Location)

```python
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x='Location', y='Acquisition_Cost', ci=None,color='skyblue')
plt.title('Acquisition Cost by Location')
plt.show()
```

# Bar Chart (Conversion_Rate by Location categorized by Target_Audience)

```python
plt.figure(figsize=(12, 6))
sns.barplot(data=df, x='Location', y='Conversion_Rate', hue='Target_Audience', ci=None)
plt.title('Conversion Rate by Location & Target Audience')
plt.show()
```

**Pie Chart (Proportion of ROI by Location)**

```python
roi_by_location = df.groupby('Location')['ROI'].sum()
plt.figure(figsize=(7, 7))
plt.pie(roi_by_location, labels=roi_by_location.index, autopct='%1.1f%%')
plt.title('Proportion of ROI by Location')
plt.show()
```

# **Key Insights & Recommendations Summary**

**Channel Performance Analysis**

Marketing channels (Facebook, Google Ads, Instagram, Website, Email, YouTube) maintain a highly consistent average Conversion Rate across the board.

The Total ROI distribution across channels is balanced, showing that each channel contributes steadily to overall campaign performance.

 Campaign & Customer Segment Insights **bold text**

All Campaign Types (Email, Influencer, Social Media, Display, Search) deliver similar conversion outputs without significant drops in any specific strategy.

Because performance is uniform across Customer Segments and Locations, reallocating ad spend toward lower acquisition-cost channels can optimize overall marketing efficiency.

**Metrics Relationship & Correlation**

The Scatter Plot reveals a positive relationship between Clicks and Impressions, confirming that increased visibility directly translates to higher user engagement.

Evaluating Acquisition Cost against Engagement Scores provides a clear framework for directing future budget toward the most cost-effective platforms.
